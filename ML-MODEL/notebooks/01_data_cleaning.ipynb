```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/raw/campuspulse_akgec_200k.csv")

print(df.shape)
df.head()
```

```python
missing_values = ["", " ", "NA", "N/A", "na", "null", "NULL", "None"]

df = df.replace(missing_values, np.nan)
```

```python
print("Duplicates:", df.duplicated().sum())

df = df.drop_duplicates()
```

```python
text_columns = df.select_dtypes(include=["object", "string"]).columns
for column in text_columns:
    df[column] = df[column].str.strip()
```

```python
df["complaint_date"] = pd.to_datetime(df["complaint_date"], errors="coerce")

df["complaint_time"] = pd.to_datetime(df["complaint_time"], errors="coerce").dt.time
```

```python
numeric_columns = [
    "year_of_study",
    "affected_users",
    "issue_duration_hours",
    "previous_similar_complaints",
    "complaints_last_7_days",
    "complaints_last_30_days",
    "recurrence_count",
    "equipment_age_years",
    "infrastructure_age_years",
    "estimated_repair_cost",
    "response_time_hours",
    "resolution_time_hours",
    "actual_expenditure"
]

for column in numeric_columns:
    df[column] = pd.to_numeric(df[column], errors="coerce")
```
